```python
class Countdown:
    def __init__(self, start):
        self.current = start  # Starting number

    def __iter__(self):
        return self  # Return iterator

    def __next__(self):
        if self.current < 0:
            raise StopIteration  # Stop the loop

        number = self.current
        self.current -= 1  # Decrease number
        return number


for number in Countdown(3):  # Create object
    print(number)  # Print each number
```

```python
#Generator - a simpler way to build an iterator
#It does not forget that value is extracted,  return the next value

def get_numbers():
    yield 10
    yield 20
    yield 30

numbers = get_numbers() #created a generator

print(next(numbers))
print(next(numbers))
print(next(numbers))
```

```python
def count_up_to(limit):
  number=1

  while number <=limit:
    yield number
    number +=1

for number in count_up_to(5):
    print(number)

#each yield pauses the loop - the function start from the last request
```

Decorator = wrapping a function with extra functionality

```python
def decorator(function):
    def wrapper():
        print("Bol na bhidu")
        function()
    return wrapper


def say_hello():
    user_name = input("Name: ")
    print(f"Hey {user_name}")


say_hello = decorator(say_hello)

say_hello()
```

```python
#You can use the decorator without @decorator. The @decorator syntax is just a shorter way of doing the same thing.
def decorator(function):
    def wrapper():
        print("Bol na bhidu")
        function()
    return wrapper

@decorator
def say_hello():
    user_name = input("Name: ")
    print(f"Hey {user_name}")



say_hello()
```

```python
#*args - accepts all positional arguments , *kwargs - it accepts all keyword arguments

def decorator(function):
  def wrapper(*args,**kwargs):
    result=function(*args,**kwargs)     #pass all arguments
    return result
  return wrapper

@decorator
def add(a,b):
  return a+b

add(10,20)
```

```python
import time
from functools import wraps


def show_time(function):
    @wraps(function)  # Keeps the original function information
    def wrapper(*args, **kwargs):
        start = time.time()  # Start the timer

        result = function(*args, **kwargs)  # Run the original function

        print(f"Time: {time.time() - start:.4f}s")

        return result  # Return result

    return wrapper


@show_time  # decorator that measures and prints the execution time of another function.
def train_one_epoch():
    # Code that performs one training epoch
    time.sleep(2)
    return "Training completed"


print(train_one_epoch())
```

```python
# property decorator - method behave like an attribute, so we dont need ().
class Student:
    def __init__(self, name):
        self.name = name

    @property
    def student_name(self):
        return self.name


student = Student("Sahil")

print(student.student_name) #without @property we would have needed ()
```

```python
# staticmethod - no need of self
class Calculator:

    @staticmethod
    def add(a, b):
        return a + b


print(Calculator.add(10, 20))
```

```python
#specific pytorch decorator, used when no need to calculate gradients and does prediction
@torch.no_grad()
def predict(model, data):
    return model(data)
```

Context Manager - setups when entered a block and clean up when leaving.

```python
#without context manager
file = open("data.txt")

content = file.read()

file.close()
#if there is error before closing file , the file is not closed properly.
```

```python
#with context manager
with open("data.txt") as file:
    content = file.read()
#even if there is a error ,clean up is handled
```

```python
class MyContext:

    def __enter__(self):   #entering the with block
        print("Entering")
        return self

    def __exit__(self, exc_type, exc_value, tb): #exiting the with block
        print("Leaving")


with MyContext():
    print("Inside")
    x=10/0


#exc_type  -What type of error?
#exc_value -What error/message?
#tb        -Where did it happen?(traceback info)
```

```python
import time
from contextlib import contextmanager


@contextmanager  #A decorator used to create context managers easily using a generator
def timer():
    start = time.time()  # Start timer (setup)

    try:
        yield            # Run the with-block

    finally:  #(cleanup)
        #  calculate the time
        print(f"Time: {time.time() - start:.4f}s")


with timer():
    time.sleep(2)
```
