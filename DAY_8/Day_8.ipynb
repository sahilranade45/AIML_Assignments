Q1 — Even or Odd Counter

20 min · loops, conditionals, comprehensions
You're given a hardcoded list of numbers:
numbers = [12, 7, 9, 4, 22, 15, 8, 3, 30, 11]

Write a script that:

Loops through the list and prints whether each number is even or odd

Counts how many are even and how many are odd, using simple counters

Uses a comprehension to build a new list containing only the even

Prints the final counts and the even-numbers list

```python
numbers=[12,7,9,4,22,15,8,3,30,11]

even_no=0
odd_no=0
even_list=[]

for i in numbers:
    if i%2==0:
        print(f"{i} is even")          #printing whether it is odd or even
        even_no=even_no+1
        even_list.append(i)            #adding even numbers in another list
    else:
        print(f"{i} is odd")
        odd_no=odd_no+1

print(f"{even_no} numbers are even")    #no of even numbers
print(f"{odd_no} numbers are odd")      #no of odd numbers
print("Even number list : ",even_list)  #printing of even numbers list

```

Q2 — Safe Average Calculator
20 min · functions, type hints, docstrings, try/except

Write a function calculate_average(numbers: list) -> float that returns the average of a list of numbers, with a type hint and a docstring.

Then write a small driver that:

Tests the function on a normal list, e.g. [10, 20, 30]

Tests it on an empty list [] — this should raise a ZeroDivisionError since you can't divide by zero

Wraps each call in a try/except that catches that specific exception and prints a friendly message instead of crashing

```python
def calculate_average(number : list) -> float :       #function definition
    return sum(number) / len(number)                  #returns average

try:
  print("Average : ",calculate_average([10,20,30]))   #function calling
except ZeroDivisionError:
  print("Empty list")

try:
  print("Average : ",calculate_average([]))   #Exception handling
except ZeroDivisionError:
  print("Empty list")
```

Q3 — Number Guessing Stats
20 min · random module, loops, math module

Simulate the computer guessing a random number between 1 and 50, ten times, using the random module.

Use random.randint(1, 50) to generate each of the 10 "guesses"

Store all 10 guesses in a list, printing each one as it's generated

After the loop, use the math module to print the highest guess (math.floor won't be needed here — just use max()) and the average guess
rounded to 2 decimal places

```python
import random       #imported random,math module
import math

guess_list=[]       #declared a empty list for adding guess in it

for i in range(10):           #10 guesses limit
  guess=random.randint(1,50)  #taking random numbers from 1 to 50
  guess_list.append(guess)    #adding the guess to empty list
  print("Guess is : ",guess)  #printing the guess

highest=max(guess_list)       #highest number of guess from list
print("Highest Guess is : ",highest)

avg = sum(guess_list) / len(guess_list)  #average of guess
print("Average Guess is : ",round(avg,2))  #rounding the average to 2 decimal place
```

Q4 — Student Class
30 min · classes, __init__, type hints, one dunder method

Design a Student class:

__init__(self, name: str, marks: list) — both type-hinted instance

A method average(self) -> float that returns the average of the student's marks

__str__ — returns a readable one-line summary, e.g. "Sakshi: average 82.5"

Then create a hardcoded list of 3 Student objects, print each one (this should use your __str__ automatically), and print the name of
whichever student has the highest average.

```python
class Student:                                 #Student class
    def __init__(self,name:str,marks:list):    #init function with name and marks
      self.name=name
      self.marks=marks

    def average(self) ->float:                #method for average
      return sum(self.marks)/len(self.marks)  #returning average

    def __str__(self):
      return f"{self.name} : Average {round(self.average(),2)}"  #returning name and average

student1=Student("Sahil",[99,78,86])  #three objects
student2=Student("Rohit",[76,88,79])
student3=Student("Omkar",[96,80,75])

students=[student1,student2,student3] #storing it in list

for std in students:
  print(std)          #displaying name with average

highest_std = student1   #storing student1 in highest_std

for std in students:
  if std.average() > highest_std.average():  #comparing each student with student1 and if student1 is smaller
    highest_std=std                          #change highest_std

print("Highest Average : ",highest_std.name)   #print highest average student name

```
