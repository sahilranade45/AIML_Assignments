```python
#In If else the condition is checked and if satisfied it executes and terminates.
spam=0
if spam<5:
  print("Hello world")
  spam=spam+1
```

```python
#Whereas in while it checks the condition and executes block of code if satisfied , and again checks the condition again n again till it satisfies.
spam=0
while spam<5:
  print("Hello world")
  spam=spam+1
```

```python
name=''
while name!='your name':
  print("Please type your name")
  name=input('>')
print("Thank you!")
```

```python
while True:
  print("Type your name")
  name=input(">")
  if name=="your name":
    break
  print("Thank you!")
```

```python
for num in range(10):
  if num%2==0:
    continue
  print(num)
```

```python
#MINI PROJECT

# This is a guess the number game.
import random
secret_number = random.randint(1, 20)

print('I am thinking of a number between 1 and 20.')

# Ask the player to guess 6 times.
for guesses_taken in range(1, 7):
    print('Take a guess.')
    guess = int(input('>'))
    if guess < secret_number:
        print('Your guess is too low.')
    elif guess > secret_number:
        print('Your guess is too high.')
    else:
        break # This condition is the correct guess!

if guess == secret_number:
    print('Good job! You got it in ' + str(guesses_taken) + ' guesses!')
else:
    print('Nope. The number was ' + str(secret_number))
```
