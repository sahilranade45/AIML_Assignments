```python
# ZeroDivisionError

def divide(n):
    try:
        return 42 / n
    except ZeroDivisionError:
        print("Error: Cannot divide by zero")


print(divide(2))
print(divide(10))
print(divide(0))
print(divide(4))

```

```python
# ValueError

def convert_number(value):
    try:
        return int(value)
    except ValueError:
        print("Error: Invalid value")


print(convert_number("10"))
print(convert_number("hey"))
```

```python
# TypeError

def add_values(a, b):
    try:
        return a + b
    except TypeError:
        print("Error: Mismatched types")


print(add_values(10, 20))
print(add_values("10", 20))
```

```python
# IndexError

def get(index):
    try:
        numbers = [1, 2]
        return numbers[index]
    except IndexError:
        print("Error: Index out of range")


print(get(0))
print(get(7))
```
