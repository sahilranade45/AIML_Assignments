```python
#Exception Handling - It means continuing remaining program even after a failure in between code
def divide(n):
  return 42/n

print(divide(2))
print(divide(10))
print(divide(0))
print(divide(4))
```

```python
def divide(n):
    try:
        return 42 / n
    except ZeroDivisionError:
        print("Error: Cannot divide by zero")


print(divide(2))
print(divide(10))
print(divide(0))
print(divide(4))
```

```python
#MINI PROJECT
#this is animation program
import time, sys
indent = 0 # How many spaces to indent
indent_increasing = True # Whether the indentation is increasing or not
try:
    while True: # The main program loop
        print(' ' * indent, end='')
        print('********')
        time.sleep(0.1) # Pause for 1/10th of a second.
        if indent_increasing:
            # Increase the number of spaces:
            indent = indent + 1
    if indent == 20:
        # Change direction:
        indent_increasing = False
    else:
        # Decrease the number of spaces:
        indent = indent - 1
    if indent == 0:
        # Change direction:
        indent_increasing = True
except KeyboardInterrupt:
    sys.exit()
```
