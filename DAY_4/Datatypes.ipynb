**STRING DATATYPE**

```python
string1=str(input("Enter a string : "))

string2=str(input("Enter a string : "))

print(string1+string2)            #String Concatenation

print(string1*2)                  #String Replication

print(string1[2])                   #String Indexing

print(string1[2:5])                 #String Slicing

print(string1.upper())              #String Uppercase

print(string2.lower())             #String Lowercase

name = "hey whatsup"

print(name.replace("hey","hello"))  #String Replace

print(len(name))                    #String Length
```

**LIST**

```python
list = [1, 17, 'hi', 'BMW', 1.6, {1: "Boy"}]

print(list[3])              # List Indexing

print(list[0:3])            # List Slicing

list.append(75)
print(list)                 # Adding an element at last

list.remove(17)
print(list)                 # Removing first matching element

list.insert(4, 10)
print(list)                 # Adding element at specific location

list.pop(4)
print(list)                 # Removing element at specific location


numbers = [5, 2, 8, 1, 10]

numbers.sort()
print(numbers)              # Sorting the list


numbers.reverse()
print(numbers)              # Reverse list


print(len(list))            # Length of list
```

**TUPLE**

```python
numbers = (10,20,10,30,40,50,40)

print(numbers[0])         #Tuple Indexing

print(numbers[1:4])       #Tuple Slicing

print(numbers.count(40))  #Occurences of specific elements

print(numbers.index(30))  #Position of element
```

**DICTIONARY**

```python
student = {"name": "Sahil","age": 20,"branch": "Computer"}

print(student.keys())       #Print Keys

print(student.values())     #Print Values

print(student.items())      #Print Items

print(student.get("name"))  #Print value of specific key

print(student.pop("age"))   #Remove Key and return its value
```

**SET**

```python
s = {1, 2, 3}

s.add(4)
print(s)                 # Adding an element to the set

s.remove(2)
print(s)                 # Removing an element from the set

s2 = {3, 4, 5}

print(s | s2)            # Union of two sets
print(s & s2)            # Intersection of two sets
```

**DATATYPE** **CHECKING**

```python
print(type(42))          # Integer
print(type(3.14))        # Float
print(type("hi"))        # String
print(type([1, 2]))      # List
print(type((1, 2)))      # Tuple
print(type({"a": 1}))    # Dictionary
print(type({1, 2}))      # Set
```

**BOOLEAN**

```python
print(bool(1))       # Converts 1 to True

print(bool(0))       # Converts 0 to False

print(10 > 5)        # Greater than

print(10 == 5)       # Equal to
```
