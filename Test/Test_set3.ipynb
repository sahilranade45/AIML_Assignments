```python
#importing pipeline from transformers , a hugging face function
from transformers import pipeline

#creating a summarization pipeline
summarizer = pipeline("summarization")

text = """
Artificial intelligence is becoming an important technology in many industries.
Large language models can understand and generate human-like text.
These models are being used for chatbots, coding, document processing, and automation.
Efficient inference helps these applications respond faster while reducing computing costs.
"""

summary = summarizer(text, max_length=50, min_length=15, do_sample=False)

print("Summary:")
print(summary[0]["summary_text"])
```

```python
from transformers import pipeline

# using text generation model gpt -2
generator = pipeline("text-generation", model="gpt2")

# continuation prompt
prompt = "My favorite hobby is"

# Generate 20 tokens
output_20 = generator(prompt,max_new_tokens=20,do_sample=True)

print("20 tokens:")
print(output_20[0]["generated_text"])
```

```python
# Generate 50 tokens
output_50 = generator(prompt,max_new_tokens=20,do_sample=True)

print("50 tokens:")
print(output_50[0]["generated_text"])
```

Generating 20 tokens has shorter continuation , but 50 tokens has more details and continuation.
