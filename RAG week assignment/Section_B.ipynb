```python
def create_chunks(text, chunk_size=100, overlap=20): #creating a function for creating a chunks having size 100 and a overlap of 20 words.
    words = text.split()  #using split function to convert sentences into single separate words having spaces
    chunks = []

    start = 0

    while start < len(words):           # continue creating chunks while words are remaining.
        end = start + chunk_size
        chunk = " ".join(words[start:end])
        chunks.append(chunk)

        start += chunk_size - overlap           # Move to the next chunk while having the last 20 words of last chunk.

    return chunks


all_text = " ".join(documents)

chunks = create_chunks(all_text)

print("Number of chunks:", len(chunks))

for i, chunk in enumerate(chunks):
    print("\nChunk", i + 1)
    print(chunk)
```

I selected a chunk size of 100 words to maintain a balance between retrieving  information and preserving sufficient context.  I used a 20-word overlap so that important information near the boundary of two chunks is not lost. This helps improve the chances of retrieving complete and meaningful information during RAG retrieval.

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

model = SentenceTransformer("all-MiniLM-L6-v2")  # Load the pre-trained embedding model

chunk_embeddings = model.encode(chunks)   # convert all the text chunks into  embeddings

def dense_retrieve(query, top_k=3):
    query_embedding = model.encode([query])   # convert the user's query into an embedding

    scores = cosine_similarity(query_embedding,chunk_embeddings)[0]  # find the cosine similarity betwwen query and chunk embeddings

    top_indices = scores.argsort()[-top_k:][::-1]

    for index in top_indices:
        print("\nScore:", scores[index])
        print("Chunk:", chunks[index])
```

```python
query1 = "What is artificial intelligence?"

dense_retrieve(query1)
```

```python
query2 = "How does RAG retrieve information?"

dense_retrieve(query2)
```

```python
from rank_bm25 import BM25Okapi
```

```python
tokenized_chunks = [chunk.lower().split() for chunk in chunks] # Convert each chunk into lowercase words

bm25 = BM25Okapi(tokenized_chunks) # Create the BM25 retrieval model
```

```python
def bm25_retrieve(query, top_k=3):
    query_tokens = query.lower().split()      # Convert the query into lowercase words

    scores = bm25.get_scores(query_tokens)        # Calculate BM25 relevance scores for all chunks

    top_indices = scores.argsort()[-top_k:][::-1]

    # Display the results
    for index in top_indices:
        print("\nScore:", scores[index])
        print("Chunk:", chunks[index])
```

```python
query1 = "What is artificial intelligence?"

bm25_retrieve(query1)
```

```python
query2 = "How does RAG retrieve information?"

bm25_retrieve(query2)
```

Dense retrieval mainly looks at the semantic meaning whereas the sparse retrieval looks at the exact word matching.
